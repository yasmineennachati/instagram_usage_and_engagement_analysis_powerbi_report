```python
import pandas as pd
```

```python
from sqlalchemy import create_engine
```

```python
# Lire le CSV Kaggle
df = pd.read_csv('instagram_usage_lifestyle.csv', encoding='utf-8')
```

```python
# Ajouter les colonnes techniques
df['batch_id'] = 1
df['source_file'] = 'instagram_usage_lifestyle.csv'
```

```python
pip install pyodbc
```

```python
import urllib
```

```python

# Construire chaîne ODBC correctement
params = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    r"SERVER=LAPTOP-2ERPMS5U\SQLEXPRESS;"
    "DATABASE=Instagram_Staging;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)
```

```python
# Connexion SQL Server
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}", fast_executemany=True)
```

```python
# Charger dans STAGING
df.to_sql(
    'stg_instagram_raw',
    engine,
    if_exists='replace',
    index=False,
    chunksize=10000
)
```
